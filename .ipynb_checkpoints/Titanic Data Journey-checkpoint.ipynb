# Prepare for analysis

## Load packages

```python
from pathlib import Path
from IPython.display import display

from titanic_q4.data import (
    load_data, preview_data, show_schema, describe_data,
    missing_data_summary, most_frequent_summary, unique_values_summary,
    combine_train_test, title_survival_summary,
)
from titanic_q4.features import (
    add_family_size, add_age_interval, add_fare_interval, add_sex_pclass,
    parse_names, add_name_features, add_family_type, standardize_titles, encode_sex,
)
from titanic_q4.plots import plot_count_grid, plot_numeric_distributions
from titanic_q4.model import (
    split_train_validation, prepare_model_data,
    fit_predict_random_forest, classification_summary,
)
```

## Process names


When we process names, we would like to extract the following information:

- Family name - this is the first word (or few first words, if a family name with multiple names), followed by a comma  
- Title - this follows just after the comma   
- Given name - this is the word or group of words following family name  
- Maiden name - for ladies, is given between parantheses  

We start with creating a function that parses the Name string and extract (if possible) these 4 elements


```python
# parse_names is now defined in features.py. Demonstrate it on one passenger.
parse_names(train_df.iloc[0])
```

# Baseline model

## Feature engineering: Map categorical value to numerical values

```python
train_df = encode_sex(train_df)
test_df = encode_sex(test_df)
```

Create train-validation split.

```python
train, valid = split_train_validation(train_df, valid_size=VALID_SIZE,
                                      random_state=RANDOM_STATE)
```

Define predictor features and target feature.

```python
print("Predictor features:", PREDICTORS)
print("Target feature:", TARGET)
```

Define the training and validation data and labels.

```python
train_X, train_Y, valid_X, valid_Y = prepare_model_data(
    train, valid, predictors=PREDICTORS, target=TARGET
)
```

Initialize the classifiction algorithm.

```python
# One function initialises/fits the Random Forest and makes both predictions.
clf, preds_tr, preds = fit_predict_random_forest(
    train_X, train_Y, valid_X,
    n_estimators=N_ESTIMATORS, random_state=RANDOM_STATE,
    criterion=CRITERION, n_jobs=N_JOBS,
)
```

The function above fits the Random Forest once, then returns predictions for both training and validation observations.

# Evaluation

## Classification report for training data.

```python
print(classification_summary(train_Y, preds_tr))
```

Classification report for validation data.

```python
print(classification_summary(valid_Y, preds))
```
